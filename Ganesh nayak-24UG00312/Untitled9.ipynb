Q1)The digital root R is calculated from the roll number.
The critical points of the function are R-1 and R+1.
These values represent where system behavior changes and will be used to define the analysis range.

```python
#Q1)
W, X, Y, Z = 1, 2, 3, 4   # replace with your digits

R = W + X + Y + Z
while R > 9:
    R = sum(int(d) for d in str(R))

R

x1 = R - 1
x2 = R + 1

print("Critical Points:", x1, x2)


```

The critical points obtained from the function represent positions where system behavior changes. These values will be used to define the range for further analysis, guiding how the dataset is examined in the next stages.

ACT 4: THE CITY AS A SIGNAL
Analysis of Aggregated Patterns in NYC Taxi Data

```python
# Convert datetime
df_clean['tpep_pickup_datetime'] = pd.to_datetime(df_clean['tpep_pickup_datetime'])

# Extract features
df_clean['hour'] = df_clean['tpep_pickup_datetime'].dt.hour
df_clean['day'] = df_clean['tpep_pickup_datetime'].dt.day_name()
df_clean['date'] = df_clean['tpep_pickup_datetime'].dt.date
```

```python
hourly_trips = df_clean.groupby('hour').size()
```

```python
hourly_fare = df_clean.groupby('hour')['fare_amount'].mean()
```

```python
daily_trips = df_clean.groupby('date').size()
```

```python
#Visualization #Hourly Trip Pattern
import matplotlib.pyplot as plt

plt.figure()
hourly_trips.plot()
plt.title("Trip Volume by Hour")
plt.xlabel("Hour")
plt.ylabel("Trips")
plt.show()
```

```python
plt.figure()
hourly_fare.plot()
plt.title("Average Fare by Hour")
plt.xlabel("Hour")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure()
daily_trips.plot()
plt.title("Trips per Day")
plt.xlabel("Date")
plt.ylabel("Trips")
plt.show()
```

1)Data Transformation:-
The dataset was aggregated by hour and day to analyze system-level behavior such as trip volume and average fare.

2)Observed Patterns
1. Patterns that Persist
*Clear peaks during morning and evening hours
*Consistent demand across weekdays
*Repeating daily cycle

3)Patterns that Disappear:-
Random spikes seen in raw data disappear after aggregation
Noise reduces when grouped by time

4)Patterns that Were Never Real:-
Extreme values (outliers) created false spikes
These were artifacts of individual trips, not system behavior

5)Interpretation (City Rhythm)
The system reveals a cyclical pattern:

*High activity during commuting hours
*Low activity during late night

This indicates a stable and repeating urban rhythm

6)Stability vs Noise;-
Patterns repeat consistently → stable + cyclical
Not random noise
Aggregation exposes true structure

7)Final Conclusion:-
The city behaves as a structured system rather than random activity. While individual trips appear random, aggregated data reveals consistent and cyclical patterns. However, some observed structures are influenced by how the data is aggregated, highlighting the importance of careful interpretation.
